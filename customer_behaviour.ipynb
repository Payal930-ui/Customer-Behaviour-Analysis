### Save DataFrame to SQLite Database

To save the `df` DataFrame to a SQLite database, we'll use Python's built-in `sqlite3` module to create a database connection and the pandas `to_sql()` method to write the DataFrame to a table within that database.

```python
import sqlite3

# Define the SQLite database file name
sqlite_db_name = 'customer_behavior.db'

# Create a connection to the SQLite database
# If the database file doesn't exist, it will be created.
conn = sqlite3.connect(sqlite_db_name)

# Save the DataFrame to a table named 'customer_data' in the SQLite database
# if_exists='replace' will overwrite the table if it already exists
# index=False prevents pandas from writing the DataFrame index as a column in the SQL table
df.to_sql('customer', conn, if_exists='replace', index=False)

# Close the connection
conn.close()

print(f"DataFrame successfully saved to SQLite database '{sqlite_db_name}' in table 'customer'.")
```

You can verify the data has been written by reading it back into a new DataFrame:

```python
# Verify by reading the data back from SQLite
conn = sqlite3.connect(sqlite_db_name)
df_from_sqlite = pd.read_sql_query("SELECT * FROM customer ", conn)


display(df_from_sqlite.head(20))
```

# **Q1. What is the total revenue generated by the male vs  the female customers?**

```python
df_from_sqlite=pd.read_sql_query("SELECT gender,SUM(purchase_amount) AS revenue  FROM customer GROUP BY gender",conn)
display(df_from_sqlite.head())
```

# **Q2. Which customer used a discount but still spent more than the averagge purchase amount?**

```python
df_from_sqlite=pd.read_sql_query("SELECT customer_id,purchase_amount from customer where discount_applied='Yes' AND purchase_amount > (SELECT AVG(purchase_amount) FROM customer)",conn)
display(df_from_sqlite.head())
```

# **Which are the top 5 products with the highest average review rating ?**

```python
df_from_sqlite=pd.read_sql_query("SELECT item_purchased,ROUND(AVG(review_rating),2) AS avg_rating FROM customer GROUP BY item_purchased ORDER BY avg_rating DESC LIMIT 5",conn)
display(df_from_sqlite.head())
```

# **Q3. Compare the average purchase  ammount between standard and express shipping.**

```python
df_from_sqlite=pd.read_sql_query("SELECT shipping_type, ROUND(AVG(purchase_amount),2) AS average_purchase_amt FROM customer WHERE shipping_type IN ('Standard','Express') GROUP BY shipping_type",conn)
display(df_from_sqlite.head())
```

# **Q5. Do subscribed customers spend more? Compare average and total revenue of subscriber and non subscribers.**

```python
df_from_sqlite=pd.read_sql_query("SELECT subscription_status,COUNT(customer_id)AS total_customers, ROUND(AVG(purchase_amount),2) AS avg_spend,ROUND(SUM(purchase_amount),2)AS total_revenue FROM customer GROUP BY subscription_status ORDER BY total_revenue,avg_spend DESC",conn)
display(df_from_sqlite.head())
```

# **Q6. WHICH 5 products have the highest percentage of purchases with discounts applied?**

```python
df_from_sqlite=pd.read_sql_query("SELECT item_purchased, 100 * SUM(CASE WHEN discount_applied='Yes' THEN 1 ELSE 0 END)/COUNT(*) AS discount_rate FROM customer GROUP BY item_purchased ORDER BY discount_rate DESC LIMIT 5 ",conn)
display(df_from_sqlite.head())
```

# **Q7. Segment customers into new, returning and loyal based on their total no. of previous purchase,and show the count oof each segment.**

```python
df_from_sqlite=pd.read_sql_query("""WITH customer_type AS(SELECT customer_id,previous_purchases,CASE
WHEN previous_purchases=1 THEN 'New'
WHEN previous_purchases BETWEEN 2 AND 10 THEN 'Returning'
ELSE 'Loyal'
END AS customer_segment FROM customer )
SELECT customer_segment,COUNT(*) AS "No. of customers" FROM customer_type GROUP BY customer_segment""",conn)
display(df_from_sqlite.head())
```

# **Q8. What are the top 3 most purchased products with each category?**

```python
df_from_sqlite=pd.read_sql_query("""WITH item_counts AS (SELECT category,item_purchased,COUNT(customer_id) AS total_orders,
ROW_NUMBER()OVER(partition by category ORDER  BY COUNT(customer_id)DESC)  AS item_rank FROM customer GROUP BY category,item_purchased)
 SELECT item_rank,category,item_purchased,total_orders FROM item_counts WHERE item_rank<=3""",conn)
display(df_from_sqlite.head())
```

# **Q9. Are customers who are repeat buyers (more than 5 previous purchases)also likely to subscribe?**

```python
df_from_sqlite=pd.read_sql_query("""SELECT subscription_status,COUNT(customer_id) AS repeat_buyers FROM customer where previous_purchases >5 GROUP BY subscription_status""",conn)
display(df_from_sqlite.head())
```

# **Q10. What is the revenue contribution of each group?**

```python
df_from_sqlite=pd.read_sql_query("SELECT age_group,SUM(purchase_amount) AS total_revenue FROM customer GROUP BY age_group ORDER BY total_revenue DESC",conn)
display(df_from_sqlite.head())
```
